**4. Web Scrapping**

Extracting data from websites.

```python
import requests
from bs4 import BeautifulSoup

url = "https://example.com"
response = requests.get(url)

soup = BeautifulSoup(response.text, "html.parser")

titles = [t.text for t in soup.find_all("h2")]

print(titles)
```

# **Data Integration**

### **Definition**

Data integration is the process of combining data from multiple sources into a unified dataset.

Example sources:

- CRM database
- Sales database
- Marketing system
- Customer feedback

---

**Problems in Data Integration**

1. Schema mismatch
2. Duplicate records
3. Naming conflicts
4. Data inconsistency


Example:

| customer_id | name | city |
| ----------- | ---- | ---- |
| 1           | John | NY   |


Another Dataset:

| id | customer_name | location |
| -- | ------------- | -------- |
| 1  | John          | New York |


These must be **standardized before integration**


---

## **Methods of Data Integration**

**1. Concatenation (Stacking data)**

```python
import pandas as pd

df1 = pd.read_csv("sales_jan.csv")
df2 = pd.read_csv("sales_feb.csv")

df = pd.concat([df1, df2])

print(df.head())
```

**2. Merging datasets (SQL style join)**

```python
customers = pd.read_csv("customers.csv")
orders = pd.read_csv("orders.csv")

df = pd.merge(customers, orders, on="customer_id")

print(df.head())
```

Types of joins:

| Join  | Meaning           |
| ----- | ----------------- |
| inner | common records    |
| left  | all left records  |
| right | all right records |
| outer | all records       |


Example:

```python
df = pd.merge(df1, df2, on="id", how="left")
```

**3. Removing duplicates**

```python
df = df.drop_duplicates()
```

**4. Resolving naming conflicts**

```python
df.rename(columns={'cust_name': 'customer_name'}, inplace=True)
```
