## Data Manipulation with Pandas

Data manipulation refers to the process of cleaning, transforming, and modifying data to make it suitable for analysis. In Pandas, this is achieved using powerful DataFrame and Series operations.

Common data manipulation tasks include:
- Adding, removing, or renaming columns and rows  
- Handling missing values (null or NaN data)  
- Filtering and selecting subsets of data  
- Sorting and indexing data efficiently  
- Performing aggregations and transformations  

Pandas provides functions such as `drop()`, `rename()`, `fillna()`, `astype()`, and boolean indexing, which allow efficient and readable data manipulation. These operations help prepare raw datasets into a structured and analysis-ready format.

```python
# rename columns
df.rename(columns={'Transaction ID':'transaction_ID', 'Date':'sales_date', 'Product Category':'product_category','Product Name':'product_name', 'Units Sold':'units_sold', 'Unit Price':'unit_price', 'Total Revenue': 'total_revenue', 'Region':'region', 'Payment Method':'payment_method'}, inplace=True)
df.head()
```

```python
# change datatypes
df['new_revenue'] = df['total_revenue'].astype(int)
df.head()

# if there was missing value while changing datatype then fill it
# df['new_revenue'] = df['total_revenue'].fillna(df['total_revenue'].mean()).astype(int)
```

```python
df['doubled_unit_price'] = df['unit_price'].apply(lambda x:x*2)
df.head()
```

```python
# Data Agggregating and Grouping
grouped_mean = df.groupby('product_category')['total_revenue'].mean()
print(grouped_mean)
```

```python
grouped_sum = df.groupby(['product_category', 'region'])['total_revenue'].sum()
print(grouped_sum)
```

```python
# aggregate multiple functions
grouped_agg = df.groupby('region')['unit_price'].agg(['mean', 'sum', 'count'])
grouped_agg
```

```python
# merging and joining dataframes

df1 = pd.DataFrame({'Key': ['A', 'B', 'C'], 'Value1': [1,2,3]})
df2 = pd.DataFrame({'Key': ['A', 'B', 'D'], 'Value2': [4,5,6]})
```

```python
df1
```

```python
df2
```

```python
# merge DatFrame on the 'Key' columns
pd.merge(df1, df2, on='Key', how='inner') # inner join
```

```python
pd.merge(df1, df2, on='Key', how='outer') # outer join
```

```python
pd.merge(df1, df2, on='Key', how='left') # left join
```

```python
pd.merge(df1, df2, on='Key', how='right') # right join
```
