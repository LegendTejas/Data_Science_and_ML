### Reading Data From Different Sources

```python
import pandas as pd
from io import StringIO

Data = '{"employee_name": "James", "email": "james@gmail.com", "job_profile": [{"title":"Team Lead", "title2":"Sr. Developer"}]}'

df = pd.read_json(StringIO(Data))  # Read JSON string and convert it into a DataFrame
df
```

```python
df.to_json() # convert dataframe to json
```

```python
df.to_json(orient='index')
```

```python
df.to_json(orient='records')
```

```python
# read data from a url
df = pd.read_csv('https://gist.githubusercontent.com/tijptjik/9408623/raw/b237fa5848349a14a14e5d4107dc7897c21951f5/wine.csv')
df
```

```python
# convert this data from url to csv
df.to_csv('wine.csv')
```

```python
# read data from HTML file
url = "https://www.fdic.gov/bank/individual/failed/banklist.html"

df = pd.read_html(url)
df[0]
```

### What is Pickle

Pickle is a Python-specific file format designed for serializing and deserializing objects, where serialization (also called pickling) transforms Python object hierarchies into a byte stream, and deserialization (or unpickling) reverses this process, converting a byte stream back into Python objects. While the pickle format is unique to the Python ecosystem, the concepts of serialization and deserialization are common across various programming languages. Through pickling, Python objects can be converted into a format that can be stored or transmitted, and subsequently, unpickling restores these objects from the stored format back to their original state.



```python
df_excel.to_pickle('df_excel')
```

```python
pd.read_pickle('df_excel')
```
