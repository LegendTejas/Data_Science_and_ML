### Pandas-DataFrame and Series

Pandas is a powerful data manipulation library in python widely used for data analysis and data cleaning. It provides two primary data structures: Series and DataFrame. A Series is a one-dimensional array-like object, while a DataFrame is a two-dimensional, size-mutable, and potentially heterogeneous tabular data structure with labeled axes (rows and columns).

```python
import pandas as pd
data=[1,2,3,4,5]
series = pd.Series(data)
print("Series\n", series)
print(type(series))
```

```python
# create a series from dictionary
data = {'a':1, 'b':2, 'c':3}
series_dict = pd.Series(data)
print(series_dict)
```

```python
data = [10, 20, 30]
index = ['a', 'b', 'c']
pd.Series(data, index=index)
```

```python
# DataFrame

# create a dataframe from a dictionary of list
data = {
    'Name':['Tejas', 'Tom', 'Jerry'],
    'Age': [20, 80, 60],
    'City':['Bangalore', 'New York', 'Florida']
}
df1 = pd.DataFrame(data)
print(df1)
print(type(df1))
```

```python
import numpy as np
np.array(df1)
```

```python
# create a dataframe from a list of dictionary

data = [
    { 'Name': 'Tejas', 'Age': 20, 'City':'Bangalore'},
    { 'Name': 'John', 'Age': 29, 'City':'Vellore'},
    { 'Name': 'Adam', 'Age': 22, 'City':'Mangaluru'}
]

df2 = pd.DataFrame(data)
print(df2)
print(type(df2))
```

```python
# Accessing data from DataFrame
df2
```

```python
df2['Name']
```

```python
df2.loc[0] # access 1st row
```

```python
# Accessing a specified element
df2.at[1, 'Age']
```

```python
# accessing a specified element using iat
df2.iat[2, 2]
```

```python
# Data Manipulation with DataFrame
df2
```

```python
# add new column
df2['Salary'] = [50000, 60000, 40000]
df2
```

```python
# Remove the 'Salary' column
# axis=1 specifies column-wise operation (axis=0 is row-wise by default)
df2.drop('Salary', axis=1, inplace=True)
df2
```

```python
# Add +1 year to the age column
df2['Age'] = df2['Age'] + 1
df2
```
