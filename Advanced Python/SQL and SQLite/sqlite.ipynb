### SQL and SQLite

SQL (Structured Query Language) is a standard language for managing and manipulating relational databases.

SQLite is a self-contained, serverless, and zero-configuration database engine that is widely used for embedded database systems. In this lesson, we will cover the basics of SQL and SQLite, including creating databases, tables, and performing various SQL operations.

```python
import sqlite3
```

```python
# connect to a SQLite Database
connection = sqlite3.connect('example.db')
connection
```

```python
cursor = connection .cursor()
```

```python
# create a table
cursor.execute('''
    CREATE TABLE IF NOT EXISTS employees(
    id INTEGER PRIMARY KEY,
    name TEXT NOT NULL,
    age INTEGER,
    department TEXT
    )
''')

# commit the changes
connection.commit()
```

```python
cursor.execute('''
SELECT * FROM employees
''')
```

```python
# insert the data in sqlite table
cursor.execute('''
INSERT INTO employees(name, age, department)
            VALUES('Tejas', 20, 'Data Science')
''')

cursor.execute('''
INSERT INTO employees(name, age, department)
            VALUES('Ravi', 29, 'Medical')
''')

cursor.execute('''
INSERT INTO employees(name, age, department)
            VALUES('Thomas', 25, 'Finance')
''')

connection.commit()
```

```python
# query the data from the table
cursor.execute('SELECT * FROM employees')
rows = cursor.fetchall()

# print the queried data
for row in rows:
    print(row)
```

```python
# update the data in the table
cursor.execute('''
UPDATE employees
SET age = 40
WHERE name = "Thomas"
''')

connection.commit()
```

```python
cursor.execute('SELECT * FROM employees')
rows = cursor.fetchall()

for row in rows:
    print(row)
```

```python
# delete the data from the table
cursor.execute('''
DELETE FROM employees
WHERE name="Ravi"
''')

connection.commit()
```

```python
cursor.execute('SELECT * FROM employees')
rows = cursor.fetchall()

for row in rows:
    print(row)
```

```python
# working with Sales Data

# connect to an SQLite Database
connection = sqlite3.connect('sales_data.db')
cursor = connection.cursor()

# create a table for sales data
cursor.execute('''
    CREATE TABLE IF NOT EXISTS sales(
    id INTEGER PRIMARY KEY,
    date TEXT NOT NULL,
    product TEXT NOT NULL,
    sales INTEGER,
    region TEXT
    )
''')

# insert data into the sales table
sales_data = [
    ("2024-01-01", "Laptop", 1200, "North"),
    ("2024-01-02", "Mobile", 800, "South"),
    ("2024-01-03", "Tablet", 450, "East"),
    ("2024-01-04", "Laptop", 1300, "West"),
    ("2024-01-05", "Mobile", 750, "North"),
    ("2024-01-06", "Tablet", 500, "South"),
    ("2024-01-07", "Laptop", 1100, "East"),
]

cursor.executemany(
    "INSERT INTO sales (date, product, sales, region) VALUES (?, ?, ?, ?)",
    sales_data
)

connection.commit()
```

```python
cursor.execute('SELECT * FROM sales')
rows = cursor.fetchall()

for row in rows:
    print(row)
```
