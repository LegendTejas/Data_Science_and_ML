### **Iterators**

Iterators are advanced Python concpts that allow for efficient looping and memory management.
Iterators provide a way to access elements of a collection sequentially without exposing the underlying structure.

```python
lst = [1,2,3,4,5,6,7]
for i in lst:
    print(i)
```

```python
# Iterator

iterator = iter(lst)
print(type(iterator))
```

```python
# Iterate through all the element
try:
    print(next(iterator))
except StopIteration:
    print("There are no elements in the iterator")
```

```python
# String iterator
str = "Hello"
str_iterator = iter(str)

print(next(str_iterator)) # H
print(next(str_iterator)) # e
```
