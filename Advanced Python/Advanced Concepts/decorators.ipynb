### Decorators

Decorators are a powerful and flexible feature in Python that allows you to modify the behavior of a function or class method. They are commonly used to add functionability to functions or methods without modifying their actual code. This lesson covers the basics of decorators, including how to create and use them.

```python
# function copy
def welcome():
    return "Hello World"

welcome()
```

```python
wel = welcome
wel()
```

```python
del welcome # delete welcome() function
```

```python
wel() # wel() function that is a copy of welcome() function will still work
```

```python
# closures functions

def main_welcome(msg):

    def sub_welcome_method():
        print(msg)
        print("Welcome to Library")
        print("Kindly mantain silence")

    return sub_welcome_method()
```

```python
main_welcome("Hello Bro")
```

```python
def main_welcome(func, lst):
    def sub_welcome_method():
        print("Welcome to the advanced python course")
        print(func(lst))
        print("Please learn these concepts properly")
    return sub_welcome_method()
```

```python
main_welcome(len, [1,2,3,4,5])
```

```python
# decorator
def my_decorator(func):
    def wrapper():
        print("Before function execution")
        func()
        print("After function execution")
    return wrapper


@my_decorator
def say_hello():
    print("Hello, world!")


say_hello()
```

```python
def validate_positive(func):
    def wrapper(n):
        if n < 0:
            print("Error: negative number not allowed")
            return None
        return func(n)
    return wrapper


@validate_positive
def square(n):
    return n * n


print(square(5))
print(square(-3))
```

```python
# decorator with arguments

def repeat(n):
    def decorator(func):
        def wrapper(*args, **kwargs):
            for _ in range(n):
                func(*args, **kwargs)
        return wrapper
    return decorator
```

```python
@repeat(3)
def say_hello():
    print("Hello")
```

```python
say_hello()
```
