### Generators

Generators are a simpler way to create iterators. They use the yield keyword to produce a series of values lazily, which means they generate values on the fly and do not store them in memory

```python
def square(n):
    # This is a generator function because it uses 'yield' instead of 'return'.
    # Calling this function does NOT execute the loop immediately.
    # It returns a generator object that produces values one at a time.

    for i in range(3):
        # Each time 'yield' is reached:
        # 1) The current value (i**2) is returned to the caller.
        # 2) The function's state is saved.
        # 3) Execution resumes from this point on the next iteration.
        yield i**2
```

```python
square(3)
```

```python
for i in square(3):
    print(i)
```

```python
def my_generator():
    yield 1
    yield 2
    yield 3
```

```python
my_generator()
```

```python
for i in my_generator():
    print(i)
```

### Practical Example: Reading Large Files

Generators are particularly useful for reading large files because they allow you to process one line at a time without loading the entire file into memory.

```python
def read_large_file(file_path):
    with open(file_path, 'r') as file:
        for line in file:
            yield line
```

```python
file_path = 'large_file.txt'

for line in read_large_file(file_path):
    print(line.strip())
```
