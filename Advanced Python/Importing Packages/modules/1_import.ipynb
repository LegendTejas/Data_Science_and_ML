### **Importing Modules in Python**

In Python modules and packages help organize and reuse code. Here's a comprehensive guide on how to import them.

```python
import math
math.sqrt(9)
```

```python
from math import sin, pi
print(pi)
print(sin(1.57))
```

```python
import numpy as np
np.array([1,2,3,4])
```

```python
import sys          # Provides access to Python's runtime environment and module search path
import os           # Used for operating system–related operations like path handling

# Get parent folder (project_root)
ROOT_DIR = os.path.abspath(os.path.join(os.getcwd(), ".."))

# Add project_root to Python module search path
sys.path.insert(0, ROOT_DIR)

# Import function from sibling package
from package.maths import add, sub
print(add(6,5))
print(sub(6,5))
```
